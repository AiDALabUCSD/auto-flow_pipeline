# Compare AutoFlow and Human flow measurements

##### imports

```python
import pandas as pd
import numpy as np
```

### load og measurements

```python
HUMAN_FLOW_MEASUREMENTS_PATH_OG = '/maxwell-projects/Aorta_pulmonary_artery_localization/ge_testing/flow_measurements_og.csv'
df_og = pd.read_csv(HUMAN_FLOW_MEASUREMENTS_PATH_OG)
```

```python
display(df_og)
```

### load ge human measurements

```python
HUMAN_FLOW_MEASUREMENTS_PATH_GE = '/maxwell-projects/Aorta_pulmonary_artery_localization/ge_testing/flow_measurements_human_ge.csv'
df_h_ge = pd.read_csv(HUMAN_FLOW_MEASUREMENTS_PATH_GE)
```

```python
display(df_h_ge)
```

### load ge autoflow measurements

```python
AUTO_FLOW_MEASUREMENTS_PATH_GE = '/maxwell-projects/Aorta_pulmonary_artery_localization/ge_testing/flow_measurements_autoflow_ge.csv'
df_a_ge = pd.read_csv(AUTO_FLOW_MEASUREMENTS_PATH_GE)
```

```python
display(df_a_ge)
```

### merge df_a_ge and df_h_ge and then dump it

```python
df_ge = pd.merge(df_h_ge, df_a_ge, left_on='Phonetic', right_on='patient_id')
df_ge = df_ge.sort_values(by='Phonetic')
display(df_ge)
```

```python
COMPILED_FLOW_MEASUREMENTS_PATH_GE = '/maxwell-projects/Aorta_pulmonary_artery_localization/ge_testing/flow_measurements_ge.csv'
df_ge.to_csv(COMPILED_FLOW_MEASUREMENTS_PATH_GE, index=False)
```

### plot ah vs auto

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))  # Make the plot square

# Scatter plots for each y-column
plt.scatter(df_ge['AV_AH'], df_ge['Ao_auto'], color='blue',  label='Ao_auto')
plt.scatter(df_ge['AV_AH'], df_ge['AV_PR'],   color='red',   label='AV_PR')
plt.scatter(df_ge['AV_AH'], df_ge['AV_LS'],   color='green', label='AV_LS')

# Labels and legend
plt.xlabel('AV_AH')
plt.ylabel('Flow Measurements')
plt.title('Scatter Plot of AV_AH vs Aortic Valve Measurements')
plt.legend()

# Add a dashed diagonal line of slope 1
xlim = plt.xlim()
ylim = plt.ylim()
min_val = min(xlim[0], ylim[0])
max_val = max(xlim[1], ylim[1])
plt.plot([min_val, max_val], [min_val, max_val], '--', color='black')

plt.show()
```
